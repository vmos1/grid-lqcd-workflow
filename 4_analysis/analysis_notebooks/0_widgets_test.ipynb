### Pattern: imports → functions → widgets → display
Re-run only the **display cell** to refresh the plot without reloading data or rebuilding widgets.

```python
# ── imports & constants ────────────────────────────────────────────────────────
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

plt.rcParams.update({'figure.dpi': 100, 'font.size': 12})

X       = np.linspace(-2 * np.pi, 2 * np.pi, 500)
XTICKS  = [-2*np.pi, -np.pi, 0, np.pi, 2*np.pi]
XLABELS = ['-2π', '-π', '0', 'π', '2π']

FUNCS = {
    'sin(k·x)'         : lambda k: np.sin(k * X),
    'cos(k·x)'         : lambda k: np.cos(k * X),
    'sin²(k·x)'        : lambda k: np.sin(k * X) ** 2,
    'sin(k·x)·cos(k·x)': lambda k: np.sin(k * X) * np.cos(k * X),
}
COLORS = ['steelblue', 'coral', 'seagreen', 'mediumpurple']
```

```python
# ── plot function ──────────────────────────────────────────────────────────────
def draw(k, selected):
    out.clear_output(wait=True)
    if not selected:
        return
    with out:
        fig, ax = plt.subplots(figsize=(9, 4))
        for name in selected:
            color = COLORS[list(FUNCS).index(name)]
            ax.plot(X, FUNCS[name](k), lw=2, color=color, label=name)
        ax.axhline(0, color='k', lw=0.7, ls='--')
        ax.set_xlim(-2*np.pi, 2*np.pi)
        ax.set_xticks(XTICKS, labels=XLABELS)
        ax.set_xlabel('x')
        ax.set_title(f'k = {k}')
        ax.legend(loc='upper right', fontsize=10)
        ax.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
```

```python
# ── widgets ────────────────────────────────────────────────────────────────────
k_slider = widgets.IntSlider(
    value=1, min=0, max=20, step=1,
    description='k:',
    continuous_update=False,
    style={'description_width': '20px'},
    layout=widgets.Layout(width='380px'),
)

func_select = widgets.SelectMultiple(
    options=list(FUNCS),
    value=list(FUNCS)[:2],
    description='Functions:',
    rows=4,
    style={'description_width': '70px'},
    layout=widgets.Layout(width='260px'),
)

out      = widgets.Output()
controls = widgets.HBox([k_slider, func_select])

def _on_change(_):
    draw(k_slider.value, func_select.value)

k_slider.observe(_on_change, names='value')
func_select.observe(_on_change, names='value')
```

```python
# ── display (re-run this cell to refresh) ─────────────────────────────────────
display(controls, out)
draw(k_slider.value, func_select.value)
```
